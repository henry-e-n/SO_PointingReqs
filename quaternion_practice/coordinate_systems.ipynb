```python
import numpy as np
import astropy.units as u
import ephem
from scipy.constants import degree
```

```python
a, b = 0 * degree, -60 * degree
site_lat = "-22:57:30"
site_lon = "-67:47:10"
site_alt = 5200 # meters


observer = ephem.Observer()
observer.lon = site_lon
observer.lat = site_lat
observer.elevation = site_alt  # In meters
observer.epoch = "2000"
observer.temp = 0  # in Celcius
observer.compute_pressure()
```

```python
?ephem.Equatorial
```

```python
center = ephem.Equatorial(a, b, epoch="2000")
center = ephem.Equatorial(center)
print(center.ra, center.dec)

corners = []
width = 10 * degree
# Synthesize 8 corners around the center
phi = center.ra
theta = center.dec
r = width / 2
ncorner = 8
angstep = 2 * np.pi / ncorner
for icorner in range(ncorner):
    ang = angstep * icorner
    delta_theta = np.cos(ang) * r
    delta_phi = np.sin(ang) * r / np.cos(theta + delta_theta)
    patch_corner = ephem.FixedBody()
    patch_corner._ra = phi + delta_phi
    patch_corner._dec = theta + delta_theta
    corners.append(patch_corner)

center_obj = ephem.FixedBody()
center_obj._ra = center.ra
center_obj._dec = theta

center_el = np.pi / 2 - np.abs(center_obj._dec - observer.lat)
center_el_rad = center_el * u.rad
center_el_deg = center_el_rad.to(u.degree)
print(center_el_rad, center_el_deg)

site_el_max = np.pi / 2
for corner in corners:
    # print(corner._dec, observer.lat, np.abs(corner._dec - observer.lat))
    el_max = np.pi / 2 - np.abs(corner._dec - observer.lat)
    # print(el_max)
    if el_max < site_el_max:
        site_el_max = el_max

max_el_rad = site_el_max*u.rad
max_el_deg = max_el_rad.to(u.degree)
print(max_el_rad, max_el_deg)
```
