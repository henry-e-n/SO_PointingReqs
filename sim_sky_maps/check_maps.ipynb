```python
import healpy as hp
import numpy as np
import os
from astropy import units as u
```

```python
lon = 290.0 * u.deg
lat = -60.0 * u.deg
```

```python
file = "maps/sim_maps/lr_simple_map_10arcmin.fits"

if not os.path.exists(file):
    raise FileNotFoundError
loaded_map = hp.read_map(file, nest=True)
hp.gnomview(loaded_map, rot=[lon.value, lat.value], xsize=2000, nest=True)
```
