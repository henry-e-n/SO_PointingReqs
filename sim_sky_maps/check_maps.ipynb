```python
file = "maps/sim_maps/lr_simple_map_10arcmin.fits"

if not os.path.exists(file):
    raise FileNotFoundError
loaded_map = hp.read_map(file, nest=True)
hp.gnomview(loaded_map, rot=[lon.value, lat.value], xsize=200, nest=True)
```
